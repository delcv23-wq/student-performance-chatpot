```python
print("===================================")
print("   STUDENT PERFORMANCE CHATBOT")
print("===================================")

print("Welcome to the chatbot!")
```

```python
name = input("Enter your name: ")

print("Hello", name)
```

```python
mark1 = float(input("Enter mark for Subject 1: "))
mark2 = float(input("Enter mark for Subject 2: "))
mark3 = float(input("Enter mark for Subject 3: "))
mark4 = float(input("Enter mark for Subject 4: "))
mark5 = float(input("Enter mark for Subject 5: "))
```

```python
total = mark1 + mark2 + mark3 + mark4 + mark5

average = total / 5

percentage = (total / 500) * 100

print("Total Marks:", total)
print("Average:", average)
print("Percentage:", percentage, "%")
```

```python
attendance = float(input("Enter attendance percentage: "))

print("Attendance:", attendance, "%")
```

```python
if mark1 >= 40 and mark2 >= 40 and mark3 >= 40 and mark4 >= 40 and mark5 >= 40:
    result = "PASS"
else:
    result = "FAIL"

print("Result:", result)
```

```python
if percentage >= 90:
    performance = "Excellent"
elif percentage >= 75:
    performance = "Very Good"
elif percentage >= 60:
    performance = "Good"
elif percentage >= 50:
    performance = "Average"
else:
    performance = "Needs Improvement"

print("Performance:", performance)
```

```python
print("\n===================================")
print("       CHATBOT STARTED")
print("===================================")

print("You can ask:")
print("1. What is my percentage?")
print("2. What is my average?")
print("3. What is my result?")
print("4. What is my attendance?")
print("5. How is my performance?")
print("6. How can I improve?")
print("Type 'exit' to stop the chatbot.")

while True:

    question = input("\nYou: ").lower()

    if question == "exit":
        print("Bot: Thank you! Goodbye.")
        break

    elif "percentage" in question:
        print("Bot: Your percentage is", round(percentage, 2), "%")

    elif "average" in question:
        print("Bot: Your average mark is", round(average, 2))

    elif "result" in question:
        print("Bot: Your result is", result)

    elif "attendance" in question:
        print("Bot: Your attendance is", attendance, "%")

    elif "performance" in question:
        print("Bot: Your performance is", performance)

    elif "improve" in question:
        print("Bot: You can improve by studying regularly, revising lessons, and practicing previous questions.")

    elif "hello" in question or "hi" in question:
        print("Bot: Hello", name, "! How can I help you?")

    else:
        print("Bot: Sorry, I don't understand. Please ask about percentage, average, result, attendance, or performance.")
```
